````python
cancellation_mask = (
    df["InvoiceNo"].astype(str).str.upper().str.startswith("C")
)

cancellation_mask.sum()
````

````python
sales = df.loc[~cancellation_mask & (df["Quantity"] > 0) & (df["UnitPrice"] > 0)].copy()

sales.shape
````

````python
sales["line_revenue"] = sales["Quantity"] * sales["UnitPrice"] 

sales[["Quantity", "UnitPrice", "line_revenue"]].head(10)
````

````python
sales["line_revenue"].sum()
````

````python
revenue_by_country = (
    sales.groupby("Country")["line_revenue"].sum().sort_values(ascending=False)
)

revenue_by_country.head(10)
````

````python
sales["InvoiceDate"].dtype
````

````python
sales["InvoiceDate"]
````

````python
sales["month"] = sales["InvoiceDate"].dt.to_period("M")

sales[["InvoiceDate", "month"]].head(10)
````

````python
sales.groupby("month")["line_revenue"].sum().sort_index(ascending=False)
````

````python
sales.groupby("month")["InvoiceDate"].agg(["min", "max"])
````

````python
revenue_by_invoice =(
    sales.groupby("InvoiceNo")["line_revenue"].sum()
)

revenue_by_invoice.mean()
````

````python
country_summary = sales.groupby("Country").agg(revenue=("line_revenue", "sum"), invoices=("InvoiceNo", "nunique"))
country_summary.head()
````

````python
sales.columns
````

````python
# unique_invoices = sales["InvoiceNo"].nunique()
# sales["average_invoice_value"] = sales["Revenue"] / unique_invoices
# sales.head()

country_summary["average_invoice_value"] = (country_summary["revenue"] / country_summary["invoices"])
````

````python
country_summary.sort_values(
    "revenue", ascending=False
).head(5)
````

````python
overall_revenue = sales["line_revenue"].sum()
country_revenue_total = country_summary["revenue"].sum()

print(overall_revenue)
print(country_revenue_total)

country_revenue_total == overall_revenue
````
